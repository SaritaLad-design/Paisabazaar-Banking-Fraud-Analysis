## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
df1= df.copy()
````

````python
#Checking Dataset Structure
df1.info()
````

````python
df1.head()
````

````python
df1.shape
````

````python
#Checking Missing Values
df1.isnull().sum()
````

````python
#Removing Duplicate Records
df1=df1.drop_duplicates()
````

````python
#1.Does a higher annual income correlate with a higher credit score?
#Comparing Average Income by Credit Score
df1.groupby('Credit_Score')['Annual_Income'].mean().round(2)
````

````python
#2.Does the credit utilization ratio have a greater impact on credit scores compared to the total number of loans taken?
#Checking whether Credit Utilization Ratio impacts Credit Score more than Number of Loans
#Converting credit score to numeric
df1['Credit_Score_Num'] = df1['Credit_Score'].map({'Poor':1,'Standard':2,'Good':3})
#Checking correlation
df1[['Credit_Utilization_Ratio','Num_of_Loan','Credit_Score_Num']].corr()
````

````python
#3.Is having a higher number of credit cards associated with a higher credit score, assuming the credit utilization ratio is managed well?
#Filtering customers with good credit utilization(Assuming good utilization <=30)
df1_good_util = df1[df1['Credit_Utilization_Ratio']<=30]
#Comparing credit score distribution by number of credit cards
df1_good_util.groupby('Num_Credit_Card')['Credit_Score'].value_counts()
````

````python
#4.Is there a negative correlation between the number of delayed payments and credit scores?
# Calculate correlation
correlation = df1['Num_of_Delayed_Payment'].corr(df1['Credit_Score_Num']).round(2)
print("Correlation between Delayed Payments and Credit Score:", correlation)
````

### What all manipulations have you done and insights you found?



1.By **Comparing Average Income by Credit Score** we get to know that higher income is associated with better credit scores.  
2.The analysis suggests that **Credit Utilization Ratio has a greater impact on Credit Score compared to the total number of loans taken**. High credit utilization indicates higher financial risk and lower repayment capacity, which negatively affects credit ratings. On the other hand, having multiple loans does not necessarily reduce credit scores if the borrower manages repayments responsibly.

3.By **comparing credit score distribution by number of credit cards** ,it suggests that responsible credit usage and a higher credit limit can positively influence credit scores.

4.By checking the **correlation between delayed payments and credit score**, a negative correlation is derived between the number of delayed payments and credit scores.This means that as **the number of delayed payments increases, the credit score tends to decrease**.

##### 1. Why did you pick the specific chart?

A **histogram** is a graphical representation used to **show the distribution of a continuous numerical variable**. It divides the data into intervals called bins and displays how many data points fall into each bin using bars. As annual income is a **continuous numerical variable** histogram helps to visualize how income values are spread across different ranges and to detect outliers.

From the above **histogram**, we get to know that most customers have lower income range which indicates that the majority of users belong to middle-income segment.

#### Chart - 3 - Box Plot Visualization(BIVARIATE)

````python
# Chart - 2 visualization code
#Creating age group
df1['Age_Group'] = pd.cut(df1['Age'],bins= [18,25,35,45,55,65,100],labels =['18-25','26-35','36-45','46-55','56-65','65+'])
#Box plot for distribution of credit scores across age groups
plt.figure(figsize=(10,6))
sns.boxplot(x='Age_Group', y='Credit_Score', data=df1)

plt.title('Distribution of Credit Scores Across Age Groups')
plt.xlabel('Age Group')
plt.ylabel('Credit Score')
plt.show()
````

##### 1. Why did you pick the specific chart?

A **Box Plot** (also called a Box-and-Whisker Plot) is a statistical chart used to **show the distribution of numerical data**. It summarizes the data using five key values: minimum, first quartile, median, third quartile, and maximum.

1.  Shows **distribution of credit score**.

2. Displays **median, spread, and outliers**.  

1.   Helps compare **credit score variation across age groups**
